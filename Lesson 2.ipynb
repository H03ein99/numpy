# Using Numpy to work with NDArrays
## Key concepts in NDArrays:
### 1) Dimensions and Shapes
- ndim: Number of dimensions of an array(1 for a vector and 2 for a matrix)
- shape: It's a tuple that shows the length of the array in each dimension
- size: Returns the total number of elements in the array.

### 2) Reshaping
- reshape(m, n): It rearranges the array dimensions without changing the data.
- Using the number '-1':‌When you write `reshape(-1, 1)`, you are telling NumPy to calculate the number of rows based on the total number of elements while setting the number of columns to exactly 1. This is the most common technique in Scikit-learn for converting a vector into a column matrix.

```python
import numpy as np

# 1. Creating a 1D array representing the target variable (y)
y = np.array([0, 1, 0, 1, 1])
print("Target vector y:")
print("Dimensions (ndim): ", y.ndim) # Output: 1
print("Shape: ", y.shape) # Output: (5, )
print("Total elements (size): ", y.size) # Output: 5

# 2. Creating a 2D array representing a feature matrix (X)
# Shape: 3 samples, 2 features

X = np.array([
    [1.5, 2.3],
    [3.1, 4.0],
    [0.8, 1.9]
])
print("\nFeature matrix X:")
print("Dimensions (ndim):", X.ndim)       # Output: 2
print("Shape:", X.shape)                   # Output: (3, 2)
print("Total elements (size):", X.size)    # Output: 6

# 3. Handling the common Scikit-learn 1D to 2D shape requirement
# Suppose we have a single feature for 4 samples
single_feature = np.array([10, 20, 30, 40])
print("\nOriginal 1D feature shape:", single_feature.shape)  # Output: (4,)
# Reshaping to a 2D column matrix with shape (4, 1)
X_reshaped = single_feature.reshape(-1, 1)
print("Reshaped 2D matrix shape:", X_reshaped.shape)         # Output: (4, 1)
print("Updated dimensions (ndim):", X_reshaped.ndim)        # Output: 2
```

## Practice:
Suppose you have the height data for 10 patients in the form of the following one-dimensional array:

```python
heights = np.array([160, 172, 178, 165, 180, 175, 168, 182, 170, 174])
```

1. Print the ndim, shape, and size attributes of the heights array.
2. Using the `reshape` method, convert this vector into a suitable 2D matrix with the shape (10, 1) and check the new shape.
3. In the second scenario, assume that these 10 numbers actually belong to 5 individuals, with two attributes (height and weight) recorded for each person. Reshape the array to dimensions (5, 2) and print its number of dimensions (ndim).

```python
import numpy as np

# Sample dataset
heights = np.array([160, 172, 178, 165, 180, 175, 168, 182, 170, 174])

# Task1: Check initial properties
print("Initial ndim: ", heights.ndim)
print("Initial shape", heights.shape)
print("Initial elements: ", heights.size)

# Task2: Convert to a 2D feature matrix (10 samples, 1 feature)
heights_2d = heights.reshape(-1, 1)
print("\nReshaped for Scikit-learn (10, 1):")
print("New shape:", heights_2d.shape)
print("New ndim:", heights_2d.ndim)

# Task 3: Reshape into 5 samples with 2 features each
features_5x2 = heights.reshape(5, 2)
print("\nReshaped into (5, 2) matrix: ")
print(features_5x2)
print("\nReshaped into (5, 2) matrix:")
print(features_5x2)
print("Shape:", features_5x2.shape)
print("Dimensions:", features_5x2.ndim)
```
