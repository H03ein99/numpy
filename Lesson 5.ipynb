# Vectorized Operations & Broadcasting

**Why is this topic the beating heart of machine learning?**
In machine learning, we deal with massive amounts of data. If you were to use traditional Python loops (such as `for` or `while`) for tasks like feature scaling, normalization, or calculating model error, the code execution would be very slow.

NumPy offers two key concepts to address this issue:

1. **Vectorization**: Performing mathematical operations on all elements of an array simultaneously at the C-language level, without the need for Python loops.

2. **Broadcasting**: NumPy's ability to perform mathematical operations between two arrays of unequal dimensions (e.g., subtracting the mean of a specific feature from all columns of a matrix).

## Key Concepts
1. **Vectorized Operations** (Element-wise Operations)

In NumPy, all basic mathematical operators (+, -, *, /, **) are applied to each individual array element by default.

2. **Broadcasting Rule**:

Two dimensions are compatible if:

- they are equal, or

- one of them is equal to 1.

When the dimension of one array is 1, NumPy virtually replicates it to match the size of the larger array, without consuming additional memory.

3. **Statistical Aggregation Along Axes** (axis=0 and axis=1)

In Scikit-learn 2D matrices:

- axis=0: Operations are performed along the rows (vertically); i.e., statistical calculations are made for each feature (column).

- axis=1: Operations are performed along the columns (horizontally); i.e., statistical calculations are made for each sample (row).

## Code Examples

```python
import numpy as np

# 1. Element-wise operations (Vectorization vs Loops)
feature_values = np.array([10.0, 20.0, 30.0, 40.0])

# Adding a scalar (shifts all elements)
scaled_values = feature_values * 0.1
print("Scaled Values:", scaled_values)

# 2. Broadcasting a 1D array across a 2D matrix
# Matrix of 3 samples with 2 features each
X = np.array([
    [10.0, 200.0],
    [20.0, 400.0],
    [30.0, 600.0]
])

# Bias vector for the 2 features
bias = np.array([5.0, 50.0])

# Broadcasting adds 'bias' to each row of 'X'
X_adjusted = X + bias
print("\nBroadcasting Result (X + bias):")
print(X_adjusted)

# 3. Axis-based aggregations: Mean across samples (per feature)
feature_means = np.mean(X, axis=0)
feature_stds = np.std(X, axis=0)

print("\nMean of each feature (axis=0):", feature_means)
print("Standard Deviation of each feature (axis=0):", feature_stds)

# 4. Feature Standardization (Z-score Normalization) using Broadcasting
# Equation: (X - mean) / std
X_standardized = (X - feature_means) / feature_stds
print("\nStandardized Matrix (Mean ~ 0, Std ~ 1):")
print(X_standardized)
```

Problem Statement:

Assume the following feature matrix contains data for four houses with two variables (area in square meters and price in thousands of dollars):

```python
housing_data = np.array([
    [60.0, 120.0],
    [85.0, 190.0],
    [120.0, 280.0],
    [150.0, 350.0]
])
```

1. Calculate the minimum (min) and maximum (max) of each feature using `axis=0`.

2. Scale the data to the range [0, 1] (Min-Max Normalization) using broadcasting rules and the following formula: $$\text{X\_scaled} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

3. Print the scaled matrix `housing_scaled`.

4. Verify the calculations; the minimum values ​​in the new matrix should be 0.0, and the maximum values ​​should be 1.0.

```python
import numpy as np

# Feature matrix: 4 samples, 2 features [Area (sqm), Price ($k)]
housing_data = np.array([
    [60.0, 120.0],
    [85.0, 190.0],
    [120.0, 280.0],
    [150.0, 350.0]
])

# Task 1: Compute Min and Max for each feature column (axis=0)

# Task 2: Implement Min-Max Normalization using Broadcasting
# Shape alignment: (4, 2) - (2,) -> Broadcasted smoothly

# Task 3: Display normalized result

# Task 4: Verification of minimum and maximum boundaries
```

## Answer is

```python
import numpy as np

# Feature matrix: 4 samples, 2 features [Area (sqm), Price ($k)]
housing_data = np.array([
    [60.0, 120.0],
    [85.0, 190.0],
    [120.0, 280.0],
    [150.0, 350.0]
])

print("Original Feature Matrix:\n", housing_data)

# Task 1: Compute Min and Max for each feature column (axis=0)
col_min = np.min(housing_data, axis=0)
col_max = np.max(housing_data, axis=0)

print("\nFeature Minimums:", col_min)
print("Feature Maximums:", col_max)

# Task 2: Implement Min-Max Normalization using Broadcasting
# Shape alignment: (4, 2) - (2,) -> Broadcasted smoothly
housing_scaled = (housing_data - col_min) / (col_max - col_min)

# Task 3: Display normalized result
print("\nMin-Max Scaled Matrix (Values in range [0, 1]):\n", housing_scaled)

# Task 4: Verification of minimum and maximum boundaries
print("\nVerification Minimums:", np.min(housing_scaled, axis=0))
print("Verification Maximums:", np.max(housing_scaled, axis=0))
```
