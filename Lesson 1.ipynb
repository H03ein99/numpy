# dtype and numerical data types

## Create elementary array

```python
import numpy as np
X = np.array([1,2,3,4])
print(X.dtype) # output: int64 or int32
```

## Convert to suitable floating‌ point number for Scikit-learn

```python
X_float = X.astype(np.float32)
print(X_float.dtype) #‌ output: float32
```

### create large data with float64

```python
large_data = np.ones((1000, 1000), dtype=np.float64)
print(large_data.nbytes) #Size of consuming memory in Bytes
```

### Lets reduce it to half

```python
optimized_data = large_data.astype(np.float32)
print(optimized_data.nbytes) #Now we see that the consuming memory has reduced in half
```
